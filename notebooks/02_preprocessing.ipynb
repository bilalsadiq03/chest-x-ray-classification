```python
from pathlib import Path

PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / "data"
EXTRACT_DIR = DATA_DIR / "raw" / "chest_xray"
SPLIT_DIR = PROJECT_DIR / "outputs" / "splits"

print("Project:", PROJECT_DIR)
print("Dataset exists:", EXTRACT_DIR.exists())

for split in ["train", "val", "test"]:
    for label in ["NORMAL", "PNEUMONIA"]:
        folder = EXTRACT_DIR / split / label

        count = (
            sum(1 for p in folder.iterdir() if p.is_file())
            if folder.exists() else 0
        )

        print(f"{split}/{label}: {count}")
```
